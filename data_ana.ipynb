Constantly a difference, more prominent in devices 15 and 17. Difference between different phs seems more or less (in terms of area), to be the same instead of the ph7 when the difference gets accentuated -> need to do analysis to see if statistical significant differences. 

```python
import os
import warnings  
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import linregress, ttest_rel # Added ttest_rel for the paired t-test
warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")

def plot_hysteresis_area_vs_ph_with_stats(data_folder, file_mapping, devicetitle):
    V_GATE_COL = "GateV"
    I_DRAIN_COL = "DrainI"
    
    ph_points = []
    auc_diff_points = []
    
    # Lists to hold the raw pairs for the statistical test
    all_auc_forward = []
    all_auc_backward = []
    
    print(f"\n--- Statistical & Calibration Results for {devicetitle} ---")

    for filename, ph_val in file_mapping.items():
        file_path = os.path.join(data_folder, filename)
        
        if not os.path.exists(file_path):
            print(f"Warning: File {filename} not found. Skipping.")
            continue
    
        df = pd.read_excel(file_path)
        
        midpoint = int(len(df) / 2)
        df_forward = df.iloc[:midpoint]
        df_backward = df.iloc[midpoint:]

        df_forward = df_forward[df_forward[I_DRAIN_COL] > 0]
        df_backward = df_backward[df_backward[I_DRAIN_COL] > 0]

        if df_forward.empty or df_backward.empty:
            continue

        df_f_sorted = df_forward.sort_values(by=V_GATE_COL)
        df_b_sorted = df_backward.sort_values(by=V_GATE_COL)

        # Using the updated trapezoid function
        auc_forward = np.trapezoid((df_f_sorted[I_DRAIN_COL]), df_f_sorted[V_GATE_COL])
        auc_backward = np.trapezoid((df_b_sorted[I_DRAIN_COL]), df_b_sorted[V_GATE_COL])

        auc_diff = auc_backward - auc_forward
        
        ph_points.append(ph_val)
        auc_diff_points.append(auc_diff)
        
        # Save the raw AUC pairs for our paired t-test later
        all_auc_forward.append(auc_forward)
        all_auc_backward.append(auc_backward)
        
    if not ph_points:
        print("No valid data points extracted.")
        return

    summary_df = pd.DataFrame({"pH": ph_points, "AUC_Difference": auc_diff_points})
    summary_df = summary_df.sort_values(by="pH")
    
    # --- 1. TEST THE SLOPE SIGNIFICANCE ---
    # linregress gives us the p-value for the slope directly!
    slope, intercept, r_value, p_value_slope, std_err = linregress(summary_df["pH"], summary_df["AUC_Difference"])
    r_squared = r_value**2
    
    # --- 2. TEST THE SWEEP DIFFERENCE (HYSTERESIS) SIGNIFICANCE ---
    # Paired t-test: compares array 1 against array 2
    t_stat, p_value_hysteresis = ttest_rel(all_auc_backward, all_auc_forward)

    # --- PRINT THE STATISTICAL READOUT ---
    print("\n[ Calibration / Slope Check ]")
    print(f"  Linear Equation: AUC_Diff = ({slope:.4e}) * pH + ({intercept:.4e})")
    print(f"  R-squared (R²):  {r_squared:.4f}")
    print(f"  Slope P-value:   {p_value_slope:.4e}")
    if p_value_slope < 0.05:
        print("  -> SIGNIFICANT: The pH level reliably changes the Hysteresis.")
    else:
        print("  -> NOT SIGNIFICANT: The Hysteresis trend vs pH might just be random noise.")

    print("\n[ Hysteresis / Sweep Difference Check ]")
    print(f"  Sweep Difference P-value: {p_value_hysteresis:.4e}")
    if p_value_hysteresis < 0.05:
        print("  -> SIGNIFICANT: The backward sweep is reliably different from the forward sweep.")
    else:
        print("  -> NOT SIGNIFICANT: The forward and backward sweeps are statistically identical.")
    print("-" * 50)

    # Plotting (Standard stuff)
    plt.figure(figsize=(9, 7))
    plt.scatter(summary_df["pH"], summary_df["AUC_Difference"], color="crimson", s=90, zorder=3, label="Extracted Data")
    
    ph_range = np.linspace(summary_df["pH"].min() - 0.5, summary_df["pH"].max() + 0.5, 100)
    fit_line = slope * ph_range + intercept
    plt.plot(ph_range, fit_line, color="darkred", linestyle="--", linewidth=2, label="Linear Fit")

    plt.title(f"Hysteresis Calibration RFET: Area Difference vs pH\n{devicetitle}", fontsize=14, pad=15)
    plt.xlabel("pH Level", fontsize=12)
    plt.ylabel(r"AUC Difference (Backward - Forward) for $\log(I_d)$ vs $V_g$", fontsize=12)
    plt.grid(True, linestyle=":", alpha=0.6)
    
    # Added p-values to the plot's text box!
    equation_text = (f"$\\Delta \\text{{AUC}} = ({slope:.2e}) \\cdot \\text{{pH}} + ({intercept:.2e})$\n"
                     f"$R^2 = {r_squared:.4f}$\n"
                     f"Slope $p$-val = {p_value_slope:.3f}\n"
                     f"Sweep Diff $p$-val = {p_value_hysteresis:.3f}")
                     
    plt.gca().text(0.05, 0.95, equation_text, transform=plt.gca().transAxes, fontsize=11,
                   verticalalignment="top",
                   bbox=dict(boxstyle="round,pad=0.5", facecolor="white", alpha=0.85))
    
    plt.legend(loc="lower right", fontsize=10)
    plt.tight_layout()
    plt.show()
```
